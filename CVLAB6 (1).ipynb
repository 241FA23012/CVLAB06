```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Gray-level range
rmin = 100
rmax = 200

# Read image in grayscale
img = cv2.imread("/content/6in1.jpg", 0)

# Check if image was loaded
if img is None:
    print("Error: Image not found")
else:
    # Gray-level slicing without background
    slicing_wo_background = np.zeros(img.shape, dtype=np.uint8)

    # Gray-level slicing with background
    slicing_w_background = img.copy()

    # Apply gray-level slicing
    for i in range(img.shape[0]):
        for j in range(img.shape[1]):
            pixel_value = img[i][j]

            if rmin <= pixel_value <= rmax:
                slicing_wo_background[i][j] = 255
                slicing_w_background[i][j] = 255

    # Display images
    plt.figure(figsize=(15, 5))

    plt.subplot(1, 3, 1)
    plt.imshow(img, cmap='gray')
    plt.title("Original Image")
    plt.axis("off")

    plt.subplot(1, 3, 2)
    plt.imshow(slicing_wo_background, cmap='gray')
    plt.title("Without Background")
    plt.axis("off")

    plt.subplot(1, 3, 3)
    plt.imshow(slicing_w_background, cmap='gray')
    plt.title("With Background")
    plt.axis("off")

    plt.show()
```
